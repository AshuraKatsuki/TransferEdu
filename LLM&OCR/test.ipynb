```python
import mysql.connector

# Connect to MariaDB (MySQL-compatible)
conn = mysql.connector.connect(
    host="localhost",      
    user="root",          
    password="admin"  # change to your MariaDB password
)

cursor = conn.cursor()

# Create database
cursor.execute("CREATE DATABASE IF NOT EXISTS learn_mariadb;")

print("Database 'learn_mariadb' created successfully!")
```

```python
import mysql.connector

# Connect directly to the learn_mariadb database
conn = mysql.connector.connect(
    host="localhost",      
    user="root",           
    password="your_password",  
    database="learn_mariadb"
)

cursor = conn.cursor()

# Create table with LONGTEXT for large content
cursor.execute("""
CREATE TABLE IF NOT EXISTS prompt_storage (
    ID INT AUTO_INCREMENT PRIMARY KEY,
    prompt LONGTEXT NOT NULL,
    pdf_path VARCHAR(500),
    LLM_model VARCHAR(255),
    OCR_model VARCHAR(255),
    create_time TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    LLM_response LONGTEXT
);
""")

print("Table 'prompt_storage' created successfully!")

# Close connection
cursor.close()
conn.close()
```
